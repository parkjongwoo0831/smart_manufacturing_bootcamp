```python
import seaborn as sns

# =====================================================
# 4. 히스토그램 + KDE
# =====================================================

# 히스토그램 + KDE 함께 보기 (권장)
sns.histplot(df['MELT_TEMP'], bins=30, kde=True, stat='density', edgecolor='white')
plt.show()

# 클래스별 분포 비교 - KDE의 최대 장점
sns.kdeplot(data=df, x='MELT_TEMP', hue='TAG', fill=True, alpha=0.3, common_norm=False)
plt.show()
```

```python
# =====================================================
# 5. 상관관계 분석
# =====================================================
selected_col = ['MELT_TEMP', 'MOTORSPEED', 'MELT_WEIGHT', 'INSP']
print(df[selected_col].corr().round(3))

plt.figure(figsize=(4,4))
sns.heatmap(
    data=df[selected_col].corr(),
    annot=True,
    fmt = '.2f',
    linewidths=.5,
    cmap='RdYlBu_r', # Blue
    vmin=-1, vmax=1  # 컬러바 범위 -1 ~ 1
)
plt.show()
```

```python
# =====================================================
# 6. 바이올린 그래프
# =====================================================

selected_col = ['MELT_TEMP', 'MOTORSPEED', 'MELT_WEIGHT', 'INSP']

plt.figure(figsize=(14,3))

for i, col in enumerate(selected_col):
    num = 141+i
    plt.subplot(num)
    sns.violinplot(data=df, y=col, hue='TAG', hue_order=['NG', 'OK'])
    plt.title(col)

plt.tight_layout()
plt.show()
```

```python
# =====================================================
# 6. 산점도 행렬
# =====================================================

selected_col = ['MELT_TEMP', 'MOTORSPEED', 'MELT_WEIGHT', 'INSP', 'TAG']
sns.pairplot(
    df[selected_col],
    diag_kind='kde',
    hue='TAG',
    hue_order=['NG', 'OK']
)
plt.show()
```

# **회전기계 진동**

```python
# 필요한 패키지(라이브러리) 로딩
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
# =====================================================
# 1. 데이터 불러오기
# =====================================================

dir = '/content/drive/MyDrive/Colab Notebooks/AI_NGV/'
normal_ = pd.read_csv(dir+'rotor_normal.csv')
type1_ = pd.read_csv(dir+'rotor_type1.csv')
type2_ = pd.read_csv(dir+'rotor_type2.csv')
type3_ = pd.read_csv(dir+'rotor_type3.csv')

print('normal의 데이터 크기 :', normal_.shape)
print('type1의 데이터 크기 :', type1_.shape)
print('type2의 데이터 크기 :', type2_.shape)
print('type3의 데이터 크기 :', type3_.shape)
```

```python
# =====================================================
# 6. 시계열 (산점도 활용)
# =====================================================


# 초기 length 개의 시계열 가시화하는 함수
def draw_time_series(ss, length):
    plt.figure(figsize = (10, 4))
    plt.scatter(range(0,length), normal_[ss][:length], label="class = normal", marker='o', s=5)
    plt.scatter(range(0,length), type1_[ss][:length], label="class = type1", marker='o', s=5)
    plt.scatter(range(0,length), type2_[ss][:length], label="class = type2", marker='o', s=5)
    plt.scatter(range(0,length), type3_[ss][:length], label="class = type3", marker='o', s=5)

    plt.legend(loc="lower right")
    plt.xlabel("Sensor " + ss, fontsize=15)
    plt.ylabel("Sensor Value", fontsize=15)
    plt.show()
    plt.close()

# 센서 1의 그래프 가시화
draw_time_series('s1', 300)
```

```python
# 센서 1의 그래프 가시화
draw_time_series('s2', 300)
```
