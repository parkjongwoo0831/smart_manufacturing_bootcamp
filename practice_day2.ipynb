```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier, plot_tree
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report,
    confusion_matrix
)

# =====================================================
# 1. Iris 데이터 불러오기
# =====================================================
df = pd.read_csv('/content/drive/MyDrive/Colab Notebooks/AI_NGV/Iris.csv', sep=',')

# 데이터 확인
print(df.head())
print("\n데이터 크기:", df.shape)
print("\n품종별 개수:")
print(df["Species"].value_counts())
```

```python
df
```

```python
df.columns
```

```python
# =====================================================
# 2. 입력변수 X와 목표변수 y 분리
# =====================================================

selected_cols = ['SepalLengthCm', 'SepalWidthCm', 'PetalLengthCm', 'PetalWidthCm']
X = df[selected_cols]
y = df['Species']

print("\n입력 데이터 X 크기:", X.shape)
print("목표 데이터 y 크기:", y.shape)

print("\n입력 데이터 예시:")
print(X.head())

print("\n목표 데이터 예시:")
print(y.head())
```

```python
# =====================================================
# 3. 훈련셋과 테스트셋 분리
# =====================================================
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.3,          # 테스트셋 약 30%
    random_state=42,    # 결과 재현
    stratify=None               # 클래스 비율 유지
)

print("\n훈련셋 입력 데이터 크기:", X_train.shape)
print("테스트셋 입력 데이터 크기:", X_test.shape)

print("\n훈련셋 클래스 개수:")
print(y_train.value_counts())

print("\n테스트셋 클래스 개수:")
print(y_test.value_counts())
```

```python
# =====================================================
# 4. 의사결정나무 모델 생성 및 학습
# =====================================================
clf = DecisionTreeClassifier(
    criterion="gini",
    max_depth=4,
    random_state=42
)

clf.fit(X_train, y_train)

print("\n의사결정나무 학습 완료")
```

```python
# =====================================================
# 5. 모델 성능평가
# =====================================================
y_pred = clf.predict(X_test)

accuracy = accuracy_score(y_test, y_pred)

# Iris는 클래스가 3개인 다중 분류 문제이므로
# precision, recall, F1 계산 시 average 지정 필요
precision_macro = precision_score(y_test, y_pred, average="macro", zero_division=0)
recall_macro = recall_score(y_test, y_pred, average="macro", zero_division=0)
f1_macro = f1_score(y_test, y_pred, average="macro", zero_division=0)


print("\n========== 모델 성능 ==========")
print(f"Accuracy       : {accuracy:.4f}")
print(f"Macro Precision: {precision_macro:.4f}")
print(f"Macro Recall   : {recall_macro:.4f}")
print(f"Macro F1-Score : {f1_macro:.4f}")


print("\n========== 분류 보고서 ==========")
print(
    classification_report(
        y_test, y_pred,
        labels=clf.classes_,
        target_names=clf.classes_,
        digits=4,
        zero_division=0
    )
)
```

```python
# =====================================================
# 6. 혼동행렬 시각화
# =====================================================
cm = confusion_matrix(
    y_test,
    y_pred,
    labels=clf.classes_
)

plt.figure(figsize=(8, 6))

sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    cmap="Blues",
    xticklabels=clf.classes_,
    yticklabels=clf.classes_
)

plt.xlabel("Predicted Species")
plt.ylabel("Actual Species")
plt.title("Iris Confusion Matrix")
plt.tight_layout()
plt.show()
```

```python
# =====================================================
# 7. 의사결정나무 시각화
# =====================================================
plt.figure(figsize=(16, 10))

plot_tree(
    clf,
    feature_names=selected_cols,
    class_names=clf.classes_,
    filled=True,
    rounded=True
)

plt.title("Iris Decision Tree")
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier, plot_tree
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report,
    confusion_matrix
)

# =====================================================
# 1. 데이터 읽기
# =====================================================
file_path = "/content/drive/MyDrive/Colab Notebooks/AI_NGV/melting_tank.csv"
df = pd.read_csv(file_path, sep=",", parse_dates=["STD_DT"])
df
```

```python
df.info()
```

```python
print("원본 데이터 크기:", df.shape)
print("\nTAG 분포:")
print(df["TAG"].value_counts(dropna=False))
```

```python
# =====================================================
# 2. TAG 전처리
# =====================================================

# 문자열 앞뒤 공백 제거 및 대문자 변환
df["TAG"] = (
    df["TAG"]
    .astype(str)
    .str.strip()
    .str.upper()
)

# OK=0, NG=1로 변환
df["TAG"] = df["TAG"].map({
    "OK": 0,
    "NG": 1
}).astype("int64")

df
```

```python
# =====================================================
# 3. 입력변수 선택 및  처리
# =====================================================

# NUM은 단순 순번이라고 가정하여 제외
feature_names = [
    "MELT_TEMP",
    "MOTORSPEED",
    "MELT_WEIGHT",
    "INSP"
]

target_name = ["TAG"]

df_model = df[feature_names + target_name]
df_model
```

```python
# =====================================================
# 4. 10,000개 비복원 추출
# =====================================================

df_sample = df_model.sample(
    n=10000,
    replace=False,
    random_state=42
).reset_index(drop=True)

print("\n표본 데이터의 클래스 분포:")
print(df_sample["TAG"].value_counts().rename(index={0: "OK", 1: "NG"}))

print("\n표본 데이터의 클래스 비율:")
print(df_sample["TAG"].value_counts(normalize=True).rename(index={0: "OK", 1: "NG"}).round(4))
```

```python
# =====================================================
# 5. 훈련셋과 테스트셋 분리
# =====================================================
X = df_sample[feature_names]
y = df_sample[target_name]

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=1/3,       # 테스트셋 약 33.3%
    random_state=42,
    stratify=y           # OK와 NG 비율 유지
)

print("\n훈련셋 크기:", X_train.shape)
print("테스트셋 크기:", X_test.shape)

print("\n훈련셋 클래스 비율:")
print( y_train.value_counts(normalize=True).rename(index={0: "OK", 1: "NG"}).round(4))

print("\n테스트셋 클래스 비율:")
print(y_test.value_counts(normalize=True).rename(index={0: "OK", 1: "NG"}).round(4))
```

```python
# =====================================================
# 6. 의사결정나무 모델 생성 및 학습
# =====================================================

clf = DecisionTreeClassifier(
    criterion="gini",
    max_depth=3,
    random_state=42
)

clf.fit(X_train, y_train)
```

```python
# =====================================================
# 7. 모델 성능평가
# =====================================================

# 테스트셋 예측
y_pred = clf.predict(X_test)

accuracy = accuracy_score(y_test, y_pred)
# NG=1을 양성 클래스로 평가
precision = precision_score(y_test, y_pred, pos_label=1, zero_division=0)
recall = recall_score(y_test, y_pred, pos_label=1,  zero_division=0)
f1 = f1_score(y_test, y_pred, pos_label=1, zero_division=0)

print("\n========== 모델 성능 ==========")
print(f"Accuracy : {accuracy:.4f}")
print(f"Precision: {precision:.4f}")
print(f"Recall   : {recall:.4f}")
print(f"F1-Score : {f1:.4f}")

print("\n========== 분류 보고서 ==========")
print(classification_report(y_test, y_pred, labels=[0, 1], target_names=["OK", "NG"], digits=4, zero_division=0))
```

```python
# =====================================================
# 8. 혼동행렬 시각화
# =====================================================
cm = confusion_matrix(y_test, y_pred, labels=[0, 1])

plt.figure(figsize=(6, 5))
sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    cmap="Blues",
    xticklabels=["OK", "NG"],
    yticklabels=["OK", "NG"]
)

plt.title("Confusion Matrix")
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.tight_layout()
plt.show()
```

```python
# =====================================================
# 9. 의사결정나무 시각화
# =====================================================
class_name_map = {
    0: "OK",
    1: "NG"
}

tree_class_names = [class_name_map[int(class_value)] for class_value in clf.classes_]

plt.figure(figsize=(16, 9))

plot_tree(
    clf,
    feature_names=list(X_train.columns),
    class_names=tree_class_names,
    filled=True,
    rounded=True
)

plt.title("Decision Tree Visualization")
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.svm import SVC
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report,
    confusion_matrix
)
```

```python
# =====================================================
# 1. Iris 데이터 불러오기
# =====================================================
iris = load_iris(as_frame=True)
df = iris.frame.copy()
df
```

```python
# =====================================================
# 2. 입력변수 및 클래스 이름 확인
# =====================================================
feature_names = list(iris.feature_names)
class_names = list(iris.target_names)

print("\n클래스 번호와 이름:")
for class_number, class_name in enumerate(class_names):
    print(f"{class_number}: {class_name}")
```

```python
# =====================================================
# 3. 입력변수 X와 목표변수 y 분리
# =====================================================

X = df[feature_names]
y = df["target"].astype("int64")

print("\n입력 데이터 X 크기:", X.shape)
print("목표 데이터 y 크기:", y.shape)

print("\n입력 데이터 예시:")
print(X.head())

print("\n목표 데이터 예시:")
print(y.head())
```

```python
# =====================================================
# 4. 훈련셋과 테스트셋 분리
# =====================================================
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=1/3,       # 테스트셋 약 33.3%
    random_state=42,     # 결과 재현
    stratify=y           # 클래스 비율 유지
)

print("\n훈련셋 입력 데이터 크기:", X_train.shape)
print("테스트셋 입력 데이터 크기:", X_test.shape)

print("\n훈련셋 클래스 개수:")
print(y_train.value_counts().sort_index().rename(index=dict(enumerate(class_names))))

print("\n테스트셋 클래스 개수:")
print(y_test.value_counts().sort_index().rename(index=dict(enumerate(class_names))))
```

```python
# =====================================================
# 5. sklearn.svm.SVC 모델 생성 및 학습
# =====================================================
svc_model = SVC(
    kernel="rbf",
    C=1.0
)

svc_model.fit(
    X_train,
    y_train
)

print("\nSVC 모델 학습 완료")
```

```python
# =====================================================
# 6. 테스트셋 예측
# =====================================================
y_pred = svc_model.predict(X_test)

class_name_map = dict(enumerate(class_names))

prediction_df = pd.DataFrame({
    "Actual": y_test.to_numpy(),
    "Predicted": y_pred
})

prediction_df["Actual_name"] = (
    prediction_df["Actual"].map(class_name_map)
)

prediction_df["Predicted_name"] = (
    prediction_df["Predicted"].map(class_name_map)
)

print("\n예측 결과 일부:")
print(prediction_df.head(10))
```

```python
# =====================================================
# 7. 전체 모델 성능평가
# =====================================================

accuracy = accuracy_score(
    y_test,
    y_pred
)

precision = precision_score(
    y_test,
    y_pred,
    average="macro",
    zero_division=0
)

recall = recall_score(
    y_test,
    y_pred,
    average="macro",
    zero_division=0
)

f1 = f1_score(
    y_test,
    y_pred,
    average="macro",
    zero_division=0
)

print("\n========== SVC 모델 성능 ==========")
print(f"Accuracy       : {accuracy:.4f}")
print(f"Macro Precision: {precision:.4f}")
print(f"Macro Recall   : {recall:.4f}")
print(f"Macro F1-Score : {f1:.4f}")
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.tree import plot_tree
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report,
    confusion_matrix
)
```

```python
# =====================================================
# 1. Iris 데이터 불러오기
# =====================================================

iris = load_iris(as_frame=True)
df = iris.frame.copy()
df
```

```python
# =====================================================
# 2. 입력변수 및 클래스 이름 확인
# =====================================================
feature_names = list(iris.feature_names)
class_names = list(iris.target_names)

print("\n클래스 번호와 이름:")
for class_number, class_name in enumerate(class_names):
    print(f"{class_number}: {class_name}")
```

```python
# =====================================================
# 3. 입력변수 X와 목표변수 y 분리
# =====================================================

X = df[feature_names]
y = df["target"].astype("int64")

print("\n입력 데이터 X 크기:", X.shape)
print("목표 데이터 y 크기:", y.shape)

print("\n입력 데이터 예시:")
print(X.head())

print("\n목표 데이터 예시:")
print(y.head())
```

```python
# =====================================================
# 4. 훈련셋과 테스트셋 분리
# =====================================================
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=1/3,       # 테스트셋 약 33.3%
    random_state=42,     # 결과 재현
    stratify=y           # 클래스 비율 유지
)

print("\n훈련셋 입력 데이터 크기:", X_train.shape)
print("테스트셋 입력 데이터 크기:", X_test.shape)

print("\n훈련셋 클래스 개수:")
print(y_train.value_counts().sort_index().rename(index=dict(enumerate(class_names))))

print("\n테스트셋 클래스 개수:")
print(y_test.value_counts().sort_index().rename(index=dict(enumerate(class_names))))
```

```python
# =====================================================
# 5. Random Forest 모델 생성 및 학습
# =====================================================

rf_model = RandomForestClassifier(
    n_estimators=100,     # 의사결정나무 100개
    criterion="gini",        # 지니 불순도 사용
    max_depth=3,          # 각 나무의 최대 깊이
    max_features="sqrt",  # 각 분할에서 일부 특징만 사용
    bootstrap=True,        # 부트스트랩 표본 사용
    oob_score=True,       # OOB 성능 계산
    random_state=42,
    n_jobs=-1             # 사용 가능한 CPU 코어 활용
)

rf_model.fit(X_train, y_train)
print("\nRandom Forest 모델 학습 완료")
```

```python
# =====================================================
# 6. 테스트셋 예측
# =====================================================

y_pred = rf_model.predict(X_test)

class_name_map = dict(enumerate(class_names))

prediction_df = pd.DataFrame({
    "Actual": y_test.to_numpy(),
    "Predicted": y_pred
})

prediction_df["Actual_name"] = (
    prediction_df["Actual"].map(class_name_map)
)

prediction_df["Predicted_name"] = (
    prediction_df["Predicted"].map(class_name_map)
)

print("\n예측 결과 일부:")
print(prediction_df.head(10))
```

```python
# =====================================================
# 7. 전체 모델 성능평가
# =====================================================

accuracy = accuracy_score(
    y_test,
    y_pred
)

precision = precision_score(
    y_test,
    y_pred,
    average="macro",
    zero_division=0
)

recall = recall_score(
    y_test,
    y_pred,
    average="macro",
    zero_division=0
)

f1 = f1_score(
    y_test,
    y_pred,
    average="macro",
    zero_division=0
)

print("\n========== RF 모델 성능 ==========")
print(f"Accuracy       : {accuracy:.4f}")
print(f"Macro Precision: {precision:.4f}")
print(f"Macro Recall   : {recall:.4f}")
print(f"Macro F1-Score : {f1:.4f}")
```

```python
# =====================================================
# 8. 클래스별 분류 보고서
# =====================================================

print("\n========== 분류 보고서 ==========")

print(
    classification_report(
        y_test,
        y_pred,
        labels=[0, 1, 2],
        target_names=class_names,
        digits=4,
        zero_division=0
    )
)
```

```python
# =====================================================
# 9. 혼동행렬 계산 및 시각화
# =====================================================

cm = confusion_matrix(
    y_test,
    y_pred,
    labels=[0, 1, 2]
)

print("\n혼동행렬:")
print(cm)

plt.figure(figsize=(7, 5))

sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    cmap="Blues",
    xticklabels=class_names,
    yticklabels=class_names
)

plt.title("Iris SVC Confusion Matrix")
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.tight_layout()
plt.show()
```

```python
# =====================================================
# 10. 변수 중요도 확인
# =====================================================

importance_df = pd.DataFrame({
    "Feature": feature_names,
    "Importance": rf_model.feature_importances_
}).sort_values(
    by="Importance",
    ascending=False
)

print("\n========== 변수 중요도 ==========")
print(importance_df)
```

```python
# =====================================================
# 11. 변수 중요도 시각화
# =====================================================

plt.figure(figsize=(8, 5))

sns.barplot(
    data=importance_df,
    x="Importance",
    y="Feature"
)

plt.title("Iris Random Forest Feature Importance")
plt.xlabel("Importance")
plt.ylabel("Feature")
plt.tight_layout()
plt.show()
```

```python
# =====================================================
# 13. 랜덤 포레스트의 첫 번째 트리 시각화
# =====================================================

first_tree = rf_model.estimators_[0]

plt.figure(figsize=(16, 9))

plot_tree(
    first_tree,
    feature_names=feature_names,
    class_names=class_names,
    filled=True,
    rounded=True
)

plt.title("The First Decision Tree in Random Forest")
plt.tight_layout()
plt.show()
```
